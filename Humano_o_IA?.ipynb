```python
import os
import nltk
import random
import collections
```

```python
nltk.download('punkt_tab')
nltk.download("stopwords")
```

```python
#Tokenizador
def tokenizar(libro):
    puntuacion = r'!"#$%&\'()*+,-./:;<=>?@[\]^_`{|}~¿¡—–'

    # Normalizar
    libro = libro.lower()

    # Tokenizar
    tokens = nltk.word_tokenize(libro, "spanish")

    # Quitar puntuación de los extremos
    for i, token in enumerate(tokens):
        tokens[i] = token.strip(puntuacion)

    # Volver a formar texto
    texto = " ".join(tokens)

    # Tokenizar nuevamente
    tokens = nltk.word_tokenize(libro, "spanish")

    # Quedarnos solamente con palabras
    tokens = [
        token for token in tokens
        if token.isalpha()]
    return tokens
```

```python
#Stop Words
palabras_funcionales=nltk.corpus.stopwords.words("spanish")
print(palabras_funcionales)
```

```python
import pandas as pd
from sklearn.feature_extraction.text import CountVectorizer
```

```python
#count vectorizer por que se nececita saber la cantidad de veces que se utilizan cada palabra
def frecuencia(texto):
  vectorizer = CountVectorizer(tokenizer=tokenizar, stop_words= palabras_funcionales)
  X = vectorizer.fit_transform(texto)
  vocabulario = vectorizer.get_feature_names_out()
  print(X.shape)

  indice = ["texto"]
  tabla_frecuencias= pd.DataFrame(X.toarray(), index=indice, columns=vocabulario)
  return tabla_frecuencias
```

```python
#Primer texto hecho por humano
tabla1 = frecuencia([ia])
tabla1
```

```python
#ordenar de mayor a menor
tabla_ordenada_1=tabla1.sort_values(by="texto", axis=1, ascending=False)
tabla_ordenada_1
```

```python
#5 plabras mas usadas
top_5 = tabla_ordenada_1.iloc[:, :5]
top_5
```

```python
#Segundo texto hecho por IA
tabla2=frecuencia([ia2])
tabla2
```

```python
#ordenar de mayor a menor
tabla_ordenada_2=tabla2.sort_values(by="texto", axis=1, ascending=False)
tabla_ordenada_2
```

```python
#5 plabras mas usadas
top_5 = tabla_ordenada_2.iloc[:, :5]
top_5
```
