- ¿Qué es la Regresión Múltiple en Redes Neuronales?  
Es un modelo que permite predecir una o varias variables apartir de varios varias variables llamadas features. Es regresión por que se regresa en el tiempo para predecir un futuro.   
  
- ¿Qué valores puede tomar el parámetro "activation" al definir una Red Neuronal? ¿Cuáles son los principales valores? ¿Cómo determinar cuál de ellos utilizar?

  Cada neurona de las capas ocultas y de salida aplican un función de activación a su suma ponderad, los valores principales son sigmoide, tanh, ReLU, softmax y identity.

  La elección de la función de activación depende del tipo de problema y de las características de los datos. Para predecir valores continuos, como ventas o precios, se puede comenzar con relu, aunque también es recomendable probar otras funciones no lineales, como tanh, para identificar cuál ofrece mejores resultados. Asimismo, es conveniente normalizar o estandarizar las variables de entrada y comparar diferentes modelos bajo las mismas condiciones, evaluando su desempeño con métricas como el error cuadrático medio (MSE), el error absoluto medio (MAE) y el coeficiente de determinación. $$(R^2) $$   

  Relu y tanh son las que se utilizan en capas ocultas.
  y logistic en cpas de salida.Sigmoide en valores binarios. softamax para multiclase (gana, pierde)


- ¿Qué valores puede tomar el parámetro "solver" al definir una Red Neuronal? ¿Cuáles son los principales valores?¿Cómo determinar cuál de ellos utilizar?
Este es para definir un algoritmo matematicos encargado de optimizar los pesos, para miminizar la función de perdida. Las principales valores son:
    - Adam: Para conjuntos medianos o grandes, y suele fucnionar bien con diferentes problemas.
    - sgd: Permite controlar el proceso de aprendizaje, necesita un configuración adecuada para que sea realmente eficiente.
    - lbfgs: Es rapido, pero es mejor solo para pocos datos.
    - rmsprop: redes recurretes de la entrada va la oculta y de nuevo a entrada

  La elección depende principalmente del tamaño del conjunto de datos, el tiempo de entrenamiento disponible y los resultados obtenidos.





##Perceptrón Multicapa

Utiliza el Perceptrón Multicapa para ajustar un modelo de Regresión para el conjunto de datos MPG que puedes cargar desde Python:



```python
import seaborn as sns

# Cargar el conjunto de datos

df = sns.load_dataset('mpg')

print(df.head())
print(df.size)
```

La variable dependiente es MPG y tu selecciona las variables independientes.   
Se tomaran como variables independientes, cylenders, displacement, horsepower, weight, acceleration y model_year.


```python
import pandas as pd
from sklearn.model_selection import train_test_split #para separa el dataset
from sklearn.preprocessing import StandardScaler #para estandarizar los datos
from sklearn.neural_network import MLPRegressor #perceptron multicapa de regresión
from sklearn.metrics import mean_squared_error, r2_score #el error
```

```python
df = df.dropna()
```

MLPRegressores una clase de la biblioteca Scikit-learn que implementa un Perceptrón Multicapa (MLP), un tipo de red neuronal artificial orientada a tareas de regresión.





Recuerda aplicar los siguientes pasos para preparar el conjunto de datos:



- Separar conjunto de datos en 70% para entrenar y 30% para probar  
- Estandarizar los conjuntos  
- Crear la instancia para MLPRegressor (define los valores para los parámetros activstion y solver de acuerdo a tu investigación)  
- Hacer el ajuste con datos train  
- Realizar pronóstico para datos de test  
- Obtener el Error Cuadrado Medio como medida de eficacia

```python
# Selección de variables independientes y dependiente
features = ['cylinders', 'displacement', 'horsepower', 'weight', 'acceleration', 'model_year']
X = df[features]
y = df['mpg']
```

```python
#Dividir los datos en conjuntos de entrenamiento (70%) y prueba (30%)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
#Estandarizar Conjuntos
#Transformar las variables independientes ($X$) para que todas tengan una media de 0 y una desviación estándar de 1 (es decir, que estén en una misma escala numérica).

from sklearn.preprocessing import StandardScaler

#Se necesita crear dos instancias, para que no se afecten las variables
scaler_x= StandardScaler()
scaler_y= StandardScaler()

# Se ajusta con los datos de entrenamiento y se transforman
X_train_scaled = scaler_x.fit_transform(X_train)
# Los de prueba solo se transforman (usando la media y desviación del entrenamiento para no hacer "trampa")
X_test_scaled = scaler_y.transform(X_test)

Y_train_scaled = scaler_y.fit_transform(y_train.values.reshape(-1, 1))
Y_test_scaled = scaler_y.transform(y_test.values.reshape(-1, 1))
```

Se debe hacer estandarización para x_train x_test y_train y_test

```python
#instancia MLPRegressor
mlp1 = MLPRegressor(
        hidden_layer_sizes=(64, 32),
        activation='relu',
        solver='adam',
        max_iter=1000,
        random_state=42
    )
```

```python
#Entrenamiento
mlp1.fit(X_train_scaled, y_train)
```

```python
#Predicción
y_pred = mlp1.predict(X_test_scaled)
```

```python
print(f"Error Cuadrático Medio (MSE): {mean_squared_error(y_test, y_pred):.4f}")
print(f"Coeficiente de Determinación (R2 Score): {r2_score(y_test, y_pred):.4f}")
```

### Nueva instancia cambiando el valor de la función de activación

```python
#instancia MLPRegressor
mlp2 = MLPRegressor(
        hidden_layer_sizes=(64, 32),
        activation='tanh',
        solver='adam',
        max_iter=1000,
        random_state=42
    )
#Entrenamiento
mlp2.fit(X_train_scaled, y_train)
#Predicción
y_pred_2 = mlp2.predict(X_test_scaled)
print(f"Error Cuadrático Medio (MSE): {mean_squared_error(y_test, y_pred_2):.4f}")
print(f"Coeficiente de Determinación (R2 Score): {r2_score(y_test, y_pred_2):.4f}")
```

```python
mlp3 = MLPRegressor(
        hidden_layer_sizes=(200, 100),
        activation='tanh',
        solver='adam',
        max_iter=1000,
        random_state=42
    )
mlp3.fit(X_train_scaled, y_train)
#Predicción
y_pred_3 = mlp3.predict(X_test_scaled)
print(f"Error Cuadrático Medio (MSE): {mean_squared_error(y_test, y_pred_3):.4f}")
print(f"Coeficiente de Determinación (R2 Score): {r2_score(y_test, y_pred_3):.4f}")
```
