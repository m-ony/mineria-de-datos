```python
#Tokenizador
def tokenizar(libro):
    puntuacion = r'!"#$%&\'()*+,-./:;<=>?@[\]^_`{|}~¿¡—–'

    # Normalizar
    libro = libro.lower()

    # Tokenizar
    tokens = nltk.word_tokenize(libro, "spanish")

    # Quitar puntuación de los extremos
    for i, token in enumerate(tokens):
        tokens[i] = token.strip(puntuacion)

    # Volver a formar texto
    texto = " ".join(tokens)

    # Tokenizar nuevamente
    tokens = nltk.word_tokenize(libro, "spanish")

    # Quedarnos solamente con palabras
    tokens = [
        token for token in tokens
        if token.isalpha()]
    return tokens
```

```python
def segmentar_oraciones_tokens(texto):
    splitter = SentenceSplitter(language = 'es')
    texto = texto.replace('\n', ' ').replace('\r', '')
    oraciones = splitter.split(texto)
    oraciones_tokenizadas = []
    for oracion in oraciones:
        if not oracion: continue
        tokens = tokenizar(oracion)
        oraciones_tokenizadas.append(tokens)
    for i, oracion in enumerate(oraciones_tokenizadas):
        oraciones_tokenizadas [i] = ["<s>"] + oracion + ["</s>"]
    return oraciones_tokenizadas
```

```python
cfd = ConditionalFreqDist(bigramas)
```

```python
entrada="la"
```

```python
probabilidades = []
for palabra, freq in cfd[entrada].items():
    prob = cfd[entrada].freq(palabra)
    probabilidades.append((palabra, prob))
  #print(f"  {palabra} (frecuencia={freq}, probabilidad={prob:.2f})")
```

```python
#Las mejores tres probabilidades
if probabilidades == []:
  print("No hay sugerencias disponibles.")
else:
  tabla_ordenada = sorted(probabilidades, key=lambda x: x[1], reverse=True)
  print("Las palabras recomendadas son:")
  print(tabla_ordenada[:3])
```
