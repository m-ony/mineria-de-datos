## Lectura, segmentación y normalización del libro

```python
#Tokenizador
def tokenizar(libro):
    puntuacion = r'!"#$%&\'()*+,-./:;<=>?@[\]^_`{|}~¿¡—–'

    # Normalizar
    libro = libro.lower()

    # Tokenizar
    tokens = nltk.word_tokenize(libro, "spanish")

    # Quitar puntuación de los extremos
    for i, token in enumerate(tokens):
        tokens[i] = token.strip(puntuacion)

    # Volver a formar texto
    texto = " ".join(tokens)

    # Tokenizar nuevamente
    tokens = nltk.word_tokenize(libro, "spanish")

    return tokens
```

```python
def segmentar_oraciones_tokens(texto):
    splitter = SentenceSplitter(language = 'es')
    texto = texto.replace('\n', ' ').replace('\r', '')
    oraciones = splitter.split(texto)
    oraciones_tokenizadas = []
    for oracion in oraciones:
        if not oracion: continue
        tokens = tokenizar(oracion)
        oraciones_tokenizadas.append(tokens)
    for i, oracion in enumerate(oraciones_tokenizadas):
        oraciones_tokenizadas [i] = [""] + oracion + [""]
    return oraciones_tokenizadas
```

```python
#Segmentación
oraciones_segmentadas = segmentar_oraciones_tokens(libro)
print(oraciones_segmentadas)
```

## Vectorizar

```python
import pandas as pd
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer(tokenizer=lambda x: x, preprocessor=lambda x: x)
X = vectorizer.fit_transform(oraciones_segmentadas)
vocabulario = vectorizer.get_feature_names_out()
print(X.shape)

indices_filas = [f"{i}" for i in range(len(oraciones_segmentadas))]
tabla_tfidf = pd.DataFrame(X.toarray(), index=indices_filas, columns=vocabulario)
tabla_tfidf

#Las veces que aprece la palabra
```

## Metrica para cada oración

Mide la importancia media de las palabras de cada oración sin beneficiar a las oraciones largas.

```python
metrica=tabla_tfidf.mean(axis=1)
metrica
```

## Ordenar las oraciones de mayor a menor

```python
tabla_tfidf['Promedio_TFIDF'] =metrica
tabla_ordenada = tabla_tfidf.sort_values(by='Promedio_TFIDF', ascending=False)
tabla_ordenada
```

```python
#Porcentaje de texto
porcentaje=int(input("¿Qué porcentaje de texto que deseas reducir? "))
numero_oraciones=int(porcentaje*len(oraciones_segmentadas)/100)
numero_oraciones
```

```python
#Armar el resumen
oraciones_seleccionadas=sorted([int(i) for i in tabla_ordenada[:numero_oraciones].index])
print(oraciones_seleccionadas)
```

```python
resumen = " \n".join(
    [" ".join(oraciones_segmentadas[i]) for i in oraciones_seleccionadas]
)
print(resumen)
```
