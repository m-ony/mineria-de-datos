Implementación de incrustaciones BERT en Python

BertTokenizer: Convierte el texto plano en tokens (palabras/subpalabras) e índices numéricos que el modelo comprende.

BertModel: Es la arquitectura del modelo BERT preentrenado que procesará los números y generará los vectores matemáticos.

torch: Framework de Deep Learning (PyTorch) que maneja los tensores y las operaciones matriciales.

```python
from transformers import BertTokenizer, BertModel
import torch

# Load pre-trained BERT model and tokenizer
tokenizer = BertTokenizer.from_pretrained('bert-base-uncased')
model = BertModel.from_pretrained('bert-base-uncased')

#Carga la versión bert-base-uncased, un modelo de tamaño base (12 capas) entrenado en texto en inglés todo en minúsculas (uncased).

# Sample text
text = "Natural Language Processing is fascinating."

# Tokenize the text
inputs = tokenizer(text, return_tensors='pt')
#Convierte la oración en una estructura de datos comprensible para PyTorch (return_tensors='pt').

# Generate BERT embeddings
with torch.no_grad():
    outputs = model(**inputs) #Contiene la salida del modelo.

# Get the embeddings for the [CLS] token (representing the entire input text)
cls_embeddings = outputs.last_hidden_state[:, 0, :]

print("BERT Embeddings for the text:")
print(cls_embeddings)
```
