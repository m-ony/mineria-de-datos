```python
#Tokenizador
def tokenizar(libro):
    puntuacion = '!"#$%&\'()*+,-./:;<=>?@[\\]^_`{|}~¿¡—–'

    # Normalizar
    libro = libro.lower()

    # Tokenizar
    tokens = nltk.word_tokenize(libro, "spanish")

    # Quitar puntuación de los extremos
    for i, token in enumerate(tokens):
        tokens[i] = token.strip(puntuacion)

    # Volver a formar texto
    texto = " ".join(tokens)

    # Tokenizar nuevamente
    tokens = nltk.word_tokenize(libro, "spanish")

    # Quedarnos solamente con palabras
    tokens = [
        token for token in tokens
        if token.isalpha()]
    return tokens
```

TF = frecuencia de la palabra \​ total de palabras del libro




```python
#Matriz TF - Frecuancia de aparición en el documento
def tf(matriz_frecuencias):
  matriz_tf = matriz_frecuencias.toarray().astype(float) # asigan los valores de la matriz de frecuencia y cambia a flotantes

  for i in range(matriz_tf.shape[0]): # for para recorrer la matriz tf

      total_palabras = matriz_tf[i].sum()  # el total de palabras en cada libro

      matriz_tf[i] = matriz_tf[i] / total_palabras
  return matriz_tf
```

```python
#Matriz IDF - En cuantos documentos aparece la palabra
def  idf (matriz_frecuencias):
  df = np.count_nonzero(
      matriz_frecuencias.toarray() > 0,
      axis=0 #por columnas
  )

  #Nos devuelve la cantidad de libros en nuetros corpus
  #shape[0] -> numero de filas
  #shape[1] -> numero de columnas

  idf = np.log(matriz_frecuencias.shape[0] / df)
  return idf
```

```python
def tf_idf(matriz_tf, idf):
  matriz_tfidf = matriz_tf * idf
  matriz_tfidf
  return matriz_tfidf
```

¿Batalla de Caballero?

```python

consulta="¿Batalla de Caballero?, Con su espada levantada y el escudo firme, luchó con valentía para proteger a su pueblo, convirtiéndose en símbolo de honor y coraje."

vectorizador_consulta = CountVectorizer( #con este ya normalizamos, limpiamos, tokenizamos
    tokenizer=tokenizar,
    stop_words=palabras_funcionales,
    vocabulary=vectorizador.vocabulary_, # le compartirmos el vocabulario aprendido
)

matriz_frecuencias_consulta = vectorizador_consulta.transform([consulta])
```

```python
matriz_frecuencias_consulta.toarray()
```

```python
#TF de consulta
total = matriz_frecuencias_consulta.sum()

tf_consulta = matriz_frecuencias_consulta/ total
tf_consulta
tf_consulta = tf_consulta.toarray()
print(tf_consulta.shape)
```

```python
tokens_consulta = tokenizar(consulta)
tokens_consulta = [
    palabra for palabra in tokens_consulta
    if palabra not in palabras_funcionales
]
tokens_consulta
```

```python
#IDF de consulta
idf_consulta = np.zeros(matriz_frecuencias.shape[1])

for palabra in tokens_consulta:
    if palabra in vectorizador.vocabulary_:
      posicion = vectorizador.vocabulary_[palabra]

      idf_consulta[posicion] = idf_corpus[posicion]

print(idf_consulta)
print(idf_consulta.shape)
```

```python
#TFIDF de Consulta
tfidf_consulta = tf_consulta * idf_consulta
print(tfidf_consulta)
print(tfidf_consulta.shape)
```
