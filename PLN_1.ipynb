```python
# Tokenizar con NLTK

import nltk
### Qué hace? Esta parte importa la biblioteca NLTK (Natural Language Toolkit), que es un conjunto de herramienta usado para procesamiento de lenguaje natural (NLP).

nltk.download('punkt_tab')  # Download the necessary resources
### Qué hace? Esta linea descargar el paquete punkt, que es necesario para realizar tareas de análisis sintáctico.

from nltk.tokenize import word_tokenize
### Qué hace? Esta linea significa que se importando la función word_tokenize del módulo nltk.tokenize de la librería NLTK (Natural Language Toolkit), la cual servira para
### segmentar texto en palabras y signos de puntuación
```

```python
# Análisis de sentimientos

from textblob import TextBlob
### Qué hace? Se esta importando la clase TextBlob de la libreria textblob, esta es una librería de procesamiento de lenguaje natural (NLP)
# que facilita el analisis de sentimiento, correcciones ortográficas, etc.

# Sample text
text = "I love this product! It's amazing."

# Create a TextBlob object
blob = TextBlob(text)
### Qué hace? Crea el objeto TextBlod apartir dde la variable text que contiene el texto para el analisis.
#Ese objeto encapsula el texto y te da acceso a varias funciones de procesamiento de lenguaje natural (PLN).

# Perform sentiment analysis
sentiment = blob.sentiment
print(sentiment)  # Output: Sentiment(polarity=0.65, subjectivity=0.6)

### Qué significa la salida? La polaridad significa que debajo de 0 toma un tono negativo y arriba de 0 toma un tono mas positivo que es lo que pasa en este caso,
# mientras tanto subjetividad arroja valores entre 0 (objetivo) y 1 (subjetivo).

### Textblob no tiene análisis de sentimientos es español
### ¿Cómo solucionarlo?
#Se podria tener el texto en español despues con translate poder cambiar alm idioma ingles y despues aplicar textblod.
### Impleméntalo con un ejemplo en español

#Declaramos la frase
frase="¡Estoy muy feliz!, Pronto es mi cumpleaños."
#Objeto traductor al ingles
translator_english= Translator(to_lang="en")
#Aplicamos la traducción al ingles
fraseTraducida=translator_english.translate(frase)
print(fraseTraducida)
#creamos el objeto Textblod y aplicamos sentiment
blob = TextBlob(fraseTraducida)
sentiment = blob.sentiment
print(sentiment)

```

```python
# Resumen de texto
!pip install sumy

from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.lsa import LsaSummarizer

### Qué hacen las funciones anteriores?
#Todas la lineas anteriores estan importanto paquetes de la libreria Sumy, la primera convierte un texto a una estructura que
#sumy pueda entender, el segundo se encarga de tokenizar el texto, es decir dividirlo en oraciones o palabras. El tercer permite genrar un resumen
#automático, analizando las relaciones semánticas entre las palabras y frases del texto.


# Sample text
text = """
Natural Language Processing (NLP) is a fascinating field.
 at the intersection of computer science,
 artificial intelligence, and linguistics.
  It enables machines to understand, interpret,
   and generate human language.
    opening up a world of possibilities. That use for applications ranging from chatbots and translation services to sentiment analysis and beyond.
"""

# Create a parser
parser = PlaintextParser.from_string(text, Tokenizer("english"))

# Create a summarizer
summarizer = LsaSummarizer()

# Generate the summary
summary = summarizer(parser.document, 2)  # Summarize to 2 sentences
for sentence in summary:
    print(sentence)


### Investigar la opción para español y hacer un ejemplo
# Para hacer una versión  en español se debe cambiar el idioma en este linea parser = PlaintextParser.from_string(text, Tokenizer("english"))

text2="""
Una niña pasaba todas las mañanas frente a un jardín lleno de flores de todos los colores. Siempre quería llevarse una flor a la escuela, así que cada día arrancaba una.
Con el tiempo, notó que las flores que tomaba se marchitaban rápidamente y que el jardín empezaba a quedarse vacío.
Al darse cuenta de que estaba acabando con la belleza del lugar, decidió dejar de arrancar flores y admirarlas desde lejos.
Desde entonces, el jardín volvió a florecer y la niña aprendió a respetar la vida y la naturaleza.
"""
parser2 = PlaintextParser.from_string(text2, Tokenizer("spanish"))

# Create a summarizer
summarizer2 = LsaSummarizer()

# Generate the summary
summary2 = summarizer2(parser2.document, 2)  # Summarize to 2 sentences
for sentence in summary2:
    print(sentence)
```

```python
## Analizar reseñas

import nltk
from nltk.sentiment.vader import SentimentIntensityAnalyzer
### Qué hace?
# Primer linea se esta importando la libreria nltlk que es un conjunto de herramientas para procesamiento de lenguaje natural.
# En la segunda linea se importa el analizador de sentimientos VADER, que es un modelo entrenado especialmente para analizar sentimientos en textos cortos e informales.

# Sample reviews
reviews = [
    "This product is fantastic! It exceeded my expectations.",
    "Not worth the price. I'm disappointed with the quality.",
    "Good value for money. Will buy again.",
]

# Initialize the sentiment analyzer
nltk.download('vader_lexicon')
### Qué hace vader_lexicon? Es un diccionario de sentimientos, es la base de datos que le da a VADER la capacidad de reconocer palabras y calcular el sentimiento de un texto.

sia = SentimentIntensityAnalyzer()

# Analyze each review
for review in reviews:
    sentiment = sia.polarity_scores(review)
    print(f"Review: {review}\\nSentiment: {sentiment}\\n")

### Qué significado tienen las salidas?
# neg es negativo el texto
# neu es neutral el texto
# pos es positivo el texto
# la suma de los tres anterios debe ser 1
# compound es el puntaje global

#En este caso no hay algo negativo en el texto, esta un 48% neutro y 51% positivo.
```

```python
# Reconocer entidades (E2)

import spacy
!python -m spacy download es_core_news_sm
### Qué hace la bilbioteca spacy?
# Permite dividir textos en palabras y oraciones, identificar categorías gramaticales,
#convertir palabras a su forma base, reconocer entidades como personas o lugares, analizar relaciones sintácticas,
#clasificar documentos, trabajar en múltiples idiomas y conectarse con frameworks de deep learning, todo con gran velocidad y eficiencia para proyectos reales.

# Load SpaCy model
nlp = spacy.load("en_core_web_sm")

### Qué hace? se esta cargando un modelo de lenguaje en ingles

text = "Google was founded by Larry Page and Sergey Brin while they were Ph.D. students at Stanford University."
doc = nlp(text)

# Extract named entities
for ent in doc.ents:
    print(ent.text, ent.label_)


### Investiga si hay spacy para español.
  #Si hay solo se necesita cargar el modelo de lenguaje en español con es_core_news_sm
### Realiza un ejemplo en español para reconocer entidades
### también aplícalo para realizar la lematización (raíces) de las palabras
### y obtener el análisis morfológico

text2 = "No se aceptan devoluciones narra la historia de Valentín, un hombre despreocupado cuya vida cambia al tener que criar solo a una hija inesperada que le dejan en la puerta."
nlp2 = spacy.load("es_core_news_sm")
doc2 = nlp2(text2)

# Extract named entities
for ent in doc2.ents:
    print(ent.text, ent.label_)
```

```python
# Clasificación con scikit-learn
from sklearn.feature_extraction.text import CountVectorizer
from sklearn.naive_bayes import MultinomialNB

### Qué es scikit-learn?
### Qué hace CountVectorizer?
### y MultinomialNB?


# Sample data
texts = ["I love this product", "This is the worst experience", "Absolutely fantastic!", "Not good at all", "I hate this", "I dislike it"]
labels = [1, 0, 1, 0, 0, 0]

# Vectorize text data
vectorizer = CountVectorizer()
X = vectorizer.fit_transform(texts)

### Qué hace?


# Train a Naive Bayes classifier
classifier = MultinomialNB()
classifier.fit(X, labels)

### Qué hace fit?

# Predict sentiment for a new text
new_text = ["I hate this"]
X_new = vectorizer.transform(new_text)
prediction = classifier.predict(X_new)
print(prediction)

### Qué significado tiene el resultado?

```
