### 🛠️ Paso 1: Configuración del Entorno e Importación de Librerías
Ejecuta la siguiente celda para asegurarte de tener todas las librerías necesarias y actualizadas en tu entorno de Google Colab.

```python
# Actualizar e importar librerías necesarias
!pip install --upgrade gensim nltk

import gensim
from gensim.models import Word2Vec
import gensim.downloader as api
import nltk
from nltk.tokenize import word_tokenize
import re

%env NLTK_ALLOW_PROXIED_URLOPEN=1
nltk.download('punkt_tab')
print("¡Librerías listas!")
```

### 📖 Paso 2: Word2Vec – Carga y Entrenamiento con 'El Quijote'
Para este paso, **debes tener cargado en tu entorno de Colab el archivo `2000.txt`** correspondientes a la obra completa. El código leerá el archivo local, limpiará el texto (removiendo signos y pasando todo a minúsculas) y entrenará el modelo Word2Vec.

```python
import os

nombre_archivo = "2000.txt" # Asegúrate de que se llame exactamente así o cámbialo

if not os.path.exists(nombre_archivo):
    raise FileNotFoundError(f"❌ ERROR: No se encontró el archivo '{nombre_archivo}'. Asegúrate de subirlo a tu panel izquierdo en Colab.")

print(f"Cargando archivo local: {nombre_archivo}...")
with open(nombre_archivo, 'r', encoding='utf-8') as f:
    texto_completo = f.read()

print("Preprocesando el texto...")
# Limpieza básica: quitar signos de puntuación y pasar a minúsculas
texto_limpio = re.sub(r'[^\w\s]', '', texto_completo.lower())

# Separar el texto en oraciones tokenizadas
oraciones = [word_tokenize(linea) for linea in texto_limpio.split('\n') if len(linea) > 5]

print("Entrenando modelo Word2Vec...")
# Tarea: Puedes experimentar modificando vector_size o window más adelante
w2v_model = Word2Vec(sentences=oraciones, vector_size=100, window=5, min_count=2, workers=4)
print("¡Modelo Word2Vec propio entrenado con éxito!")
```

### 🌍 Paso 3: GloVe – Carga de Vectores Preentrenados
Descargaremos un conjunto de vectores GloVe preentrenados globalmente que entienden español moderno para contrastarlo con el vocabulario clásico de tu modelo anterior.

```python
print("Descargando vectores preentrenados GloVe (esto puede demorar de 1 a 2 minutos)...")
glove_model = api.load("glove-twitter-25")
print("¡Vectores GloVe cargados con éxito!")
```

#### 🔍 Tarea A: Búsqueda de Similitud Semántica
Comparemos qué palabras guardan mayor relación matemática. Escribe un script completo que busque y extraiga las **3 palabras más similares** al término `"caballero"` de manera simultánea en ambos modelos.

"**Requisitos:**",
"* El código debe invocar los métodos nativos de similitud semántica de `gensim` de forma independiente para cada modelo.",

"* Debe imprimir de forma clara y ordenada los resultados (las palabras junto con su score de confianza) identificando plenamente qué salida pertenece a Word2Vec y cuál a GloVe."

```python
palabra_objetivo = "caballero"

#Búsqueda en el modelo Word2Vec
similares_w2v = w2v_model.wv.most_similar(palabra_objetivo, topn=3)

#Búsqueda en el modelo GloVe
similares_glove = glove_model.most_similar(palabra_objetivo, topn=3)

# Impresión ordenada de resultados
print(f"=== RESULTADOS DE SIMILITUD PARA: '{palabra_objetivo}' ===\n")

print("Modelo Word2Vec (Entrenado con El Quijote):")
for palabra, score in similares_w2v:
    print(f"   {palabra:<15} (Confianza: {score:.4f})")

print("\n Modelo GloVe (Preentrenado global/Twitter):")
for palabra, score in similares_glove:
    print(f"  {palabra:<15} (Confianza: {score:.4f})")
```

w2v_model.wv.most_similar(): Accede al objeto de vectores de palabras (KeyedVectors) nativo de Gensim para el modelo que se enteno localmente. Las palabras más similares reflejarán el vocabulario y contexto de la literatura.

glove_model.most_similar(): Al ser un modelo descargado directamente desde Gensim API (KeyedVectors), utiliza el método directo para consultar embeddings preentrenados globalmente.

#### 📐 Tarea B: Álgebra de Palabras (Analogías)
Intentemos resolver el clásico problema matemático vectorial: **Rey - Hombre + Mujer = Reina**.

"**Reto 2:** Desarrolla un script completo que resuelva de forma algebraica el clásico problema matemático vectorial de analogías tridimensionales: **Rey - Hombre + Mujer = Reina**."

"**Requisitos:**",

"* Tu código debe pasar los vectores positivos (`'rey'`, `'mujer'`) y el vector negativo (`'hombre'`) a las funciones correspondientes de ambos modelos."

"* Configura la salida para que solo devuelva el primer resultado con mayor probabilidad (`topn=1`).",

"* **Obligatorio:** Dado que los vocabularios de entrenamiento difieren drásticamente, debes implementar una estructura de control de excepciones (`try - except KeyError`) en ambos bloques de consulta para evitar que el script colapse si una palabra no existe en alguno de los modelos."

```python
# Términos para el cálculo vectorial: rey - hombre + mujer
positivas = ['rey', 'mujer']
negativas = ['hombre']

#positive y negative: Pasan los vectores que se suman (Rey + Mujer) y el que se resta (Hombre).

print(f"=== TAREA B: ÁLGEBRA DE VECTORES ({positivas[0]} - {negativas[0]} + {positivas[1]}) ===\n")

#Analogía en Word2Vec (El Quijote)
print("Modelo Word2Vec:")
try:
    resultado_w2v = w2v_model.wv.most_similar(positive=positivas, negative=negativas, topn=1)
    palabra, score = resultado_w2v[0]
    print(f"Resultado: '{palabra}' (Confianza: {score:.4f})")
except KeyError as e:
    print(f" Error: Al menos una de las palabras no existe en el vocabulario del modelo. Detalle: {e}")

print("\nModelo GloVe:")
# Analogía en GloVe
try:
    resultado_glove = glove_model.most_similar(positive=positivas, negative=negativas, topn=1)
    palabra, score = resultado_glove[0]
    print(f"Resultado: '{palabra}' (Confianza: {score:.4f})")
except KeyError as e:
    print(f"Error: Al menos una de las palabras no existe en el vocabulario del modelo. Detalle: {e}")
```

#### 🚪 Tarea C: Detector de Intrusos
Veamos si los modelos logran identificar cuál elemento no pertenece semánticamente al grupo.

Crea un script completo que use las propiedades geométricas de los vectores para identificar qué palabra de una lista dada no guarda coherencia ni pertenece al mismo contexto semántico que el resto.,

"**Requisitos:**",
"* Utiliza la lista de palabras fija: `['perro', 'gato', 'caballo', 'ventana']",

"* Encuentra el método nativo de exclusión semántica integrado dentro de los objetos de vectores en `gensim`."

"* Ejecuta la lógica en ambos modelos e imprime cuál es el elemento intruso determinado por Word2Vec y cuál por GloVe."

```python
# Lista de palabras para la prueba de exclusión semántica
palabras = ['perro', 'gato', 'caballo', 'ventana']

print(f"=== TAREA C: DETECTOR DE INTRUSOS ===")
print(f"Lista analizada: {palabras}\n")

# Detección en el modelo Word2Vec (El Quijote)
print("Modelo Word2Vec ")
try:
    intruso_w2v = w2v_model.wv.doesnt_match(palabras)
    print(f"El elemento intruso es: '{intruso_w2v}'")
except KeyError as e:
    print(f" Error: Al menos una palabra no existe en el vocabulario de Word2Vec ({e}).")

print("\nModelo GloVe:")
#Detección en el modelo GloVe
try:
    intruso_glove = glove_model.doesnt_match(palabras)
    print(f" El elemento intruso es: '{intruso_glove}'")
except KeyError as e:
    print(f" Error: Al menos una palabra no existe en el vocabulario de GloVe ({e}).")
```

doesnt_match(): Es el método nativo de Gensim diseñado para esta tarea. Calcula la matriz de similitud de coseno promedio entre todas las combinaciones de palabras de la lista y aísla la que tiene la menor similitud respecto al centroide geométrico del grupo.
